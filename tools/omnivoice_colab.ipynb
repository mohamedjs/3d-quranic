```
from omnivoice import OmniVoice
model = OmniVoice.from_pretrained('k2-fsa/OmniVoice', device_map='cuda:0', dtype=torch.float16)
PROMPT = {sp: model.create_voice_clone_prompt(ref_audio=f'refs/{sp}.mp3', ref_text=txt) for sp, (_, txt) in REF.items()}
print('model + voices ready')
```

```
import soundfile as sf, time
os.makedirs('voices', exist_ok=True); t0 = time.time()
for i, (id_, sp, text) in enumerate(LINES):
    out = 'voices/' + id_.replace('/', '__') + '.wav'
    if os.path.exists(out): continue
    text = text.replace('ﷺ', 'صلى الله عليه وسلم')
    try:
        audio = model.generate(text=text, voice_clone_prompt=PROMPT[sp], num_step=32)
        sf.write(out, audio[0], 24000)
        print(f'{i+1}/{len(LINES)} {id_} {len(audio[0])/24000:.1f}s  ({time.time()-t0:.0f}s)')
    except Exception as e:
        print('FAILED', id_, e)
```
